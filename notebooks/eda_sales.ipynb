```python
# Import các thư viện cần thiết
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Thiết lập phong cách biểu đồ
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

# 1. Nạp dữ liệu
df = pd.read_csv('../data/raw/supermarket_sales.csv')
df['Hour'] = pd.to_datetime(df['Time']).dt.hour

# ==========================================
# BIỂU ĐỒ 1: Khung giờ vàng mua sắm
# ==========================================
plt.figure(figsize=(10, 5))
sns.lineplot(data=df, x='Hour', y='Sales', estimator=sum, errorbar=None, marker='o', color='b')
plt.title('Tổng Doanh Thu Theo Khung Giờ Trong Ngày', fontsize=14, fontweight='bold')
plt.xlabel('Khung giờ (Hour)')
plt.ylabel('Tổng Doanh Thu ($)')
plt.xticks(range(8, 22))
plt.show()

# ==========================================
# BIỂU ĐỒ 2: Dòng sản phẩm mang lại doanh thu cao nhất
# ==========================================
plt.figure(figsize=(10, 5))
sales_by_product = df.groupby('Product line')['Sales'].sum().sort_values(ascending=False).reset_index()
sns.barplot(data=sales_by_product, y='Product line', x='Sales', palette='viridis')
plt.title('Doanh Thu Theo Dòng Sản Phẩm', fontsize=14, fontweight='bold')
plt.xlabel('Tổng Doanh Thu ($)')
plt.ylabel('Dòng Sản Phẩm')
plt.show()

# ==========================================
# BIỂU ĐỒ 3: Tỷ lệ khách hàng (Member vs Normal)
# ==========================================
plt.figure(figsize=(6, 6))
cust_type_counts = df['Customer type'].value_counts()
plt.pie(cust_type_counts, labels=cust_type_counts.index, autopct='%1.1f%%', 
        colors=['#ff9999','#66b3ff'], startangle=90, wedgeprops={'edgecolor': 'black'})
plt.title('Tỷ Lệ Tệp Khách Hàng', fontsize=14, fontweight='bold')
plt.show()
```
